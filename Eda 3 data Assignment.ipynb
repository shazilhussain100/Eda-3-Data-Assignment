```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
loan=pd.read_csv('loan-train.csv')
```

```python

loan['Self_Employed'].unique()
```

```python
loan['Gender'].fillna(loan['Gender'].mode()[0], inplace=True)
loan['Married'].fillna(loan['Married'].mode()[0],inplace=True)
loan['Dependents'].fillna(loan['Dependents'].mode()[0],inplace=True)
loan['Self_Employed'].fillna(loan['Self_Employed'].mode()[0],inplace=True)
```

```python
loan['LoanAmount'].fillna(loan['LoanAmount'].median(),inplace=True)
loan['Loan_Amount_Term'].fillna(loan['Loan_Amount_Term'].median(),inplace=True)
loan['Credit_History'].fillna(loan['Credit_History'].median(),inplace=True)
```

```python
loan.isna().sum()
```

```python
loan.dtypes
```

```python
loan['Credit_History'] =loan['Credit_History'].astype(int)
loan['Loan_Amount_Term'] =loan['Loan_Amount_Term'].astype(int)
```

```python
loan.describe()
```

```python
sns.boxplot(x=loan['LoanAmount'])
plt.show()
```

```python
sns.boxplot(x=loan['ApplicantIncome'])
plt.show()
```

```python
print("loan_data Shape:", loan.shape)
print("Missing Values:")
print(loan.isna().sum())
print("Duplicate Rows:", loan.duplicated().sum())
```

```python
loan.to_csv('Cleaned_loan_data.csv',index=False)
```
